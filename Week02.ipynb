```python
# Handle missing values

# Handle Online Backup column
df["Online Backup"].value_counts(dropna=False)
df["Online Backup"] = df["Online Backup"].fillna(df["Online Backup"].mode()[0])

# Handle Avg Monthly GB Download  column
df["Avg Monthly GB Download"].value_counts(dropna=False)
df["Avg Monthly GB Download"] = df["Avg Monthly GB Download"].fillna(df["Avg Monthly GB Download"].mean())

# Handle Avg Monthly Long Distance Charges  column
df["Avg Monthly Long Distance Charges"].value_counts(dropna=False)
df["Avg Monthly Long Distance Charges"] = df["Avg Monthly Long Distance Charges"].fillna(df["Avg Monthly Long Distance Charges"].mean())
```

##### VISUALIZATIONS

```python
# Customer Tenure Distribution

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="Tenure in Months",
    bins=20,
    kde=True
)

plt.title("Distribution of Customer Tenure")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.savefig("chart1_tenure.png", dpi=300, bbox_inches="tight")
plt.show()
```

The above chart shows how customer tenure is distributed across the telecom company's customer base. 
A histogram with a KDE curve was selected because it displays the frequency distribution and the overall shape of customer tenure.
Tenure has a U-shape: about 1,200 customers fall in the first bin (under roughly 4 months) and about 750 in the last bin (around 70 months), while the middle bins hold about 200 to 300 each. This shows a large group of very new customers and a loyal long-term base, with the mean tenure of 32.4 months sitting between them.

```python
# Churn by contract type

contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

contract_churn.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Customer Outcomes by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of Customers")
plt.legend(title="Customer Churn")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("chart2_contract_churn.png", dpi=300,
            bbox_inches="tight")
plt.show()
```

The above chart shows that the maximum joining of customers to the company's service and maximum churn is during month-to-month contract type.
Month-to-month customers churn at 45.8%, against 10.7% on one-year and 2.6% on two-year contracts. Month-to-month also has the highest share of new joiners (11.3%). Longer contracts are strongly associated with retention.

```python
# Monthly charges by churn status

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="Monthly Charge"
)

plt.title("Monthly Charges by Customer Status")
plt.xlabel("Customer Churn")
plt.ylabel("Monthly Charge")
plt.tight_layout()
plt.savefig("chart3_monthly_charges.png", dpi=300,
            bbox_inches="tight")
plt.show()
```

Churned customers have the highest median monthly charge, about 79.5, compared with about 65.6 for those who stayed and about 44 for those who joined. The churned box is also tighter at the top, with half of those customers paying between roughly 55 and 94. A cluster of negative outliers near -10 to -4 appears in the churned group, which reflects the data-quality issue 

```python
# CORRELATION Heatmap

numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(11, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Correlation Between Numerical Features")
plt.tight_layout()
plt.savefig("chart4_correlation.png", dpi=300,
            bbox_inches="tight")
plt.show()
```

The strongest relationships are between Total Charges and Total Revenue (0.97), Tenure and Total Revenue (0.85), and Tenure and Total Charges (0.83), which is expected because revenue accumulates over time. Zip Code, Latitude and Longitude are strongly related (0.89 and -0.89) because they describe location, and Age is negatively related to Avg Monthly GB Download (-0.57). The ChurnBinary row is empty because the mapping used the keys “No” and “Yes”, which do not exist in the Churn column, so no churn correlation is shown.

```python
# Revenue associated with churned customers

churned = df[df["Churn"] == "Churned"]

revenue_by_contract = (
    churned.groupby("Contract")["Total Revenue"]
    .sum()
    .sort_values()
)

plt.figure(figsize=(9, 5))

revenue_by_contract.plot(kind="barh")

plt.title("Recorded Revenue of Churned Customers by Contract")
plt.xlabel("Total Recorded Revenue")
plt.ylabel("Contract Type")
plt.tight_layout()
plt.savefig("chart5_revenue.png", dpi=300,
            bbox_inches="tight")
plt.show()
```

Month-to-Month customers represent the largest revenue segment among churned customers, making them an important target for customer retention initiatives.

```python
# Churn category among churned customer

# Filter only churned customers
churned_customers = df[df["Churn"] == "Churned"]

# Create bar chart
plt.figure(figsize=(10, 6))

ax = sns.countplot(
    data=churned_customers,
    x="Churn Category",
    order=churned_customers["Churn Category"].value_counts().index,
    palette="viridis"
)

# Add labels and title
plt.title("Churn Category Among Churned Customers",
          fontsize=15, fontweight="bold")
plt.xlabel("Churn Category", fontsize=12)
plt.ylabel("Number of Churned Customers", fontsize=12)
plt.xticks(rotation=45, ha="right")

# Display counts above bars
for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.tight_layout()
plt.show()
```

Competitor-related reasons dominate with 841 of 1,869 churned customers (45.0%), followed by dissatisfaction (321), attitude (314), price (211) and other (182). Reasons tied to service quality and support (dissatisfaction and attitude) together account for about 34% of churn, so they are a second major lever beyond competitive pricing.
