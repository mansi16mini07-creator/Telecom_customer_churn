## CLEAN AND PREPROCESS THE DATA

```python
# Count of missing values in each column
missing_counts = df.isnull().sum()
missing_counts
```

```python
# percentage of missing values in each column
missing_percentage = (df.isnull().sum() / len(df1)) * 100
missing_percentage
```

```python
# missing summary
missing_summary = pd.DataFrame({'Missing Count': missing_counts, 'Missing Percentage': missing_percentage})
missing_summary
```

```python
# Bar chart to identify which columns have the most missing observations
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

if len(missing) > 0:
    plt.figure(figsize=(10, 5))
    missing.plot(kind="bar")
    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found.")
```

```python
categorical_cols = [
    'Offer',
    'Internet Type',
    'Online Security',
    'Online Backup',
    'Device Protection Plan',
    'Premium Tech Support',
    'Streaming TV',
    'Streaming Movies',
    'Streaming Music',
    'Unlimited Data',
    'Churn Category',
    'Churn Reason'
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

```python
#revert data to original
df["Churn Category"]= df1["Churn Category"]
df["Churn Reason"]= df1["Churn Reason"]
```

```python
# Handle offer column
df["Offer"].value_counts(dropna=False)
df["Offer"] = df["Offer"].fillna("No Offer")

# Handle Multiple Lines column
#df.drop(columns=["Multiple Lines"], inplace= True)

# Handle Internet Types column
df["Internet Type"].value_counts(dropna=False)
df["Internet Type"] = df["Internet Type"].fillna(df["Internet Type"].mode()[0])

# Handle Online Security column
df["Online Security"].value_counts(dropna=False)
df["Online Security"] = df["Online Security"].fillna(df["Online Security"].mode()[0])

# Handle Device Protection Plan column
#df.drop(columns=["Device Protection Plan"])

# Handle Premium Tech Support column
#df.drop(columns=["Premium Tech Support"])

# Handle Streaming TV column
#df.drop(columns=["Streaming TV"])

# Handle Streaming Movies column
#df.drop(columns=["Streaming Movies"])

# Handle Streaming Music column
#df.drop(columns=["Streaming Music"])

# Handle Unlimited Data column
#df.drop(columns=["Unlimited Data"])

# Handle Churn Category column
df["Churn Category"].value_counts(dropna=False)
df["Churn Category"] = df["Churn Category"].fillna("unknown")

# Handle Churn Reason column
df["Churn Reason"].value_counts(dropna=False)
df["Churn Reason"] = df["Churn Reason"].fillna("unknown reason")
    
```

### PERFORM EDA AND CREATE VISUALIZATION

```python
# Churn data distibution

print(df["Churn"].value_counts(dropna=False))

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Churn")
plt.title("Distribution of Customer Churn")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
# churn percentage

churn_percentages = (
    df["Churn"].value_counts(normalize=True, dropna=False) * 100
).round(2)

display(churn_percentages)
```

```python
# Customer tenure distribution

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Tenure in Months", bins=20, kde=True)
plt.title("Distribution of Customer Tenure")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
df.columns
```

```python
# Monthly charges by churn status

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Churn", y="Monthly Charge")
plt.title("Monthly Charges by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Monthly Charges")
plt.tight_layout()
plt.show()
```

```python
# Churn by contract types

churn_by_contract = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

display(churn_by_contract.round(2))

churn_by_contract.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Churn Distribution by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of Customers")
plt.legend(title="Churn")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Exploring numerical correlations

corr_df = df.copy()

if "Churn" in corr_df.columns:
    corr_df["ChurnBinary"] = corr_df["Churn"].map({
        "No": 0,
        "Yes": 1
    })

numeric_df = corr_df.select_dtypes(include="number")

plt.figure(figsize=(12, 8))
sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```
