```python
# Data preparation

df["Contract"] = df["Contract"].astype("string").str.strip()
df["Churn"] = df["Churn"].astype("string").str.strip()

df["Monthly Charge"] = pd.to_numeric(
    df["Monthly Charge"], errors="coerce"
)

df = df.dropna(
    subset=["Contract", "Churn", "Monthly Charge"]
)

print(df[["Contract", "Churn", "Monthly Charge"]].info())
print(df["Churn"].value_counts())
print(df["Contract"].value_counts())
```

```python
# Visualize contract type and churn

plt.figure(figsize=(6, 3))

sns.countplot(
    data=df,
    x="Contract",
    hue="Churn"
)

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# chi-square test

from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

print("Contingency Table:")
print(contingency_table)

chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("P-value:", p_value)
print("Expected frequencies:\n", expected)

if (expected < 5).any():
    print("Warning: Some expected cell counts are below 5.")

if p_value < 0.05:
    print("Reject H0: Evidence of an association.")
else:
    print("Fail to reject H0: Insufficient evidence of an association.")
```

```python
# viz for chi-square test

# Create contingency table
contingency_table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

# Calculate percentages within each contract type
percentage_table = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

# Plot 100% stacked bar chart
percentage_table.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Customer Churn Percentage by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of Customers (%)")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.tight_layout()
plt.show()

# Perform Chi-square test
chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Contract type is associated with churn.")
else:
    print("Fail to reject H0: Insufficient evidence of an association.")
```

```python
# welch's test

from scipy.stats import ttest_ind

churned = df.loc[
    df["Churn"] == "Yes", "Monthly Charge"
].dropna()

retained = df.loc[
    df["Churn"] == "No", "Monthly Charge"
].dropna()

print("Churned customers:", len(churned))
print("Retained customers:", len(retained))

print("Mean charge - churned:", churned.mean())
print("Mean charge - retained:", retained.mean())

t_stat, p_value = ttest_ind(
    churned,
    retained,
    equal_var=False,
    alternative="two-sided"
)

print("Welch t-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: The group means differ significantly.")
else:
    print("Fail to reject H0: Insufficient evidence of a difference.")
```

```python
# 95% confidence interval for the difference of means


from statsmodels.stats.weightstats import (DescrStatsW, CompareMeans)

# Replace with the exact numerical column name
column = "Monthly Charge"

df[column] = pd.to_numeric(df[column], errors="coerce")

df[column] = df[column].replace(
    [np.inf, -np.inf], np.nan
)

group1 = df.loc[
    df["Churn"] == "Churned", column
].dropna()

group2 = df.loc[
    df["Churn"] == "Stayed", column
].dropna()

print("Churned mean:", group1.mean())
print("Stayed mean:", group2.mean())

if len(group1) >= 2 and len(group2) >= 2:
    mean_diff = group1.mean() - group2.mean()

    result = CompareMeans(
        DescrStatsW(group1),
        DescrStatsW(group2)
    )

    ci = result.tconfint_diff(
        usevar="unequal",
        alpha=0.05
    )

    print("Mean difference:", mean_diff)
    print("95% confidence interval:", ci)
else:
    print("Insufficient valid observations.")
```
